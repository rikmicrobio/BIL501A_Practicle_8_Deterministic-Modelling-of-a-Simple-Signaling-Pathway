## 2. What is an Ordinary Differential Equation?

An **Ordinary Differential Equation (ODE)** is a mathematical description of how a quantity changes with respect to another variable, usually **time**.

In biology, the quantity could be:

- protein concentration
- mRNA concentration
- drug concentration
- metabolite concentration
- number of cells
- concentration of a signaling molecule

The key question is:

> **How does this biological quantity change over time?**


## 3. What does `dA/dt` mean?

Consider:

$$\frac{dA}{dt}$$

Do not worry about the calculus notation at this stage.

Simply read it as:

**"How quickly is A changing with time?"**

Here:

- `A` = the biological quantity we are measuring
- `t` = time
- `dA/dt` = rate of change of A with respect to time


## 4. A Simple Everyday Example

Imagine that a water tank contains water.

- Water enters the tank.
- Water leaves the tank.

Therefore:

$$\text{Change in water} = \text{Water entering} - \text{Water leaving}$$

Biological systems can often be described in a similar way.

$$\boxed{\text{CHANGE} = \text{PRODUCTION} - \text{REMOVAL}}$$


## 5. Example: Protein Concentration

Suppose a protein is produced inside a cell and is also degraded.

We can describe its change as:

$$\text{Change in protein} = \text{Protein production} - \text{Protein degradation}$$

A simple ODE could be:

$$\frac{dP}{dt}=k_{prod}-k_{deg}P$$

where:

- `P` = protein concentration
- `k_prod` = protein production rate
- `k_deg` = protein degradation rate


## 5A. Understanding the Protein ODE

A simple ODE for protein concentration is:

$$\boxed{\frac{dP}{dt}=k_{prod}-k_{deg}P}$$

This equation describes how the protein concentration changes with time.

We can separate the equation into two biological processes:

$$\frac{dP}{dt}=\underbrace{k_{prod}}_{\text{production}}-\underbrace{k_{deg}P}_{\text{degradation}}$$

Therefore:

$$\boxed{\text{Change in protein}=\text{Production}-\text{Degradation}}$$

### What do the symbols mean?

| Symbol | Biological meaning |
|:---:|---|
| $P$ | Protein concentration |
| $k_{prod}$ | Protein production rate |
| $k_{deg}$ | Protein degradation rate constant |

### Why is there a `P` after $k_{deg}$?

The degradation term is **$k_{deg}P$**, not just $k_{deg}$.

This means that the amount of protein being degraded depends on how much protein is currently present.

For example:

$$P\uparrow \quad\Rightarrow\quad k_{deg}P\uparrow$$

In other words, when more protein is present, more protein is degraded per unit time.

This is called **first-order degradation** because the degradation term is proportional to the protein concentration $P$.

### Important terminology

It is more precise to call $k_{deg}$ the **protein degradation rate constant** rather than simply the degradation rate, because the actual degradation rate is:

$$\text{Degradation rate}=k_{deg}P$$

Similarly, $k_{prod}$ represents a constant production rate in this simplified model.

## 6. Important Terms

| Term | Meaning |
|---|---|
| Variable | A quantity that can change during the simulation |
| Time | The independent variable in our model |
| Parameter | A value that controls the behavior of the model |
| Initial condition | The starting value of a variable |
| ODE | A rule describing how a variable changes |
| Numerical solution | Values calculated by a computer at different time points |


## 7. Why Do We Need an Initial Condition?

An ODE tells us **how a quantity changes**, but we also need to know where the system starts.

For example:

$$P(0)=0$$

means:

> At time 0, the protein concentration is 0.

In Python:

```python
P0 = 0
```


## 8. A Simple Biological Signaling Example

We can now extend the idea to a simple signaling system:

```text
Signal
   |
   v
Inactive Protein
   |
   | activation
   v
Activated Protein (A)
   |
   | activates
   v
Cellular Response (R)
```

For this teaching model:

- `S` = Signal
- `A` = Activated Protein
- `R` = Cellular Response


## 9. Building the ODE from Biology

Suppose the signal promotes activation of protein `A`.

At the same time, activated protein can be deactivated or removed.

Therefore:

$$\frac{dA}{dt}=k_1S-k_2A$$

Read this as:

> **Change in activated protein = activation by signal − deactivation**

For the downstream response:

$$\frac{dR}{dt}=k_3A-k_4R$$

Read this as:

> **Change in cellular response = production by activated protein − removal**


## 10. What Do the Symbols Mean?

| Symbol | Biological meaning |
|:---:|---|
| `S` | Signal |
| `A` | Activated Protein |
| `R` | Cellular Response |
| `k1` | Rate of protein activation |
| `k2` | Rate of protein deactivation/removal |
| `k3` | Rate of cellular response production |
| `k4` | Rate of cellular response removal |

Think of these symbols simply as **short names for biological quantities or parameters**.

## 13. Import the Python Libraries

We will use:

- **NumPy** for numerical calculations
- **SciPy** for solving the ODE
- **Matplotlib** for plotting the results


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
````

## 14. A Very Simple ODE: Protein Production and Removal

Before solving the signaling model, let us start with one variable.

We will use:

$$\frac{dP}{dt}=k_{prod}-k_{deg}P$$

Let's choose simple example values:

- production rate = `1`
- degradation rate = `0.2`
- initial protein concentration = `0`


````python
k_prod = 1.0
k_deg = 0.2
P0 = [0.0]
````

## 15. Define the ODE

The function below tells Python how the protein concentration changes.

The computer will repeatedly use this rule to calculate the protein concentration at later times.


````python
def protein_model(t, y):
    P = y[0]
    dPdt = k_prod - k_deg * P
    return [dPdt]
````

## 16. Define the Time Period

Here we will simulate the system from time 0 to time 30.

We also create time points where we want Python to report the solution.


````python
t_start = 0
t_end = 30
t_eval = np.linspace(t_start, t_end, 300)
````

## 17. Solve the ODE with SciPy

`solve_ivp` is a SciPy function used to solve an initial-value ODE problem numerically.

You do not need to understand the numerical algorithm yet. For now, think of it as:

> **Give Python the ODE, the starting value, and the time range; Python calculates how the system changes over time.**


````python
solution = solve_ivp(
    protein_model,
    (t_start, t_end),
    P0,
    t_eval=t_eval
)
````

## 18. Look at the Results

The solution contains the time points and the calculated protein concentrations.


````python
time = solution.t
protein = solution.y[0]

print("First five time points:")
print(time[:5])

print("\nFirst five protein concentrations:")
print(protein[:5])
````

## 19. Plot the ODE Solution

A graph makes the time-dependent behavior easier to understand.


````python
plt.figure(figsize=(8, 5))
plt.plot(time, protein, linewidth=2)
plt.xlabel("Time")
plt.ylabel("Protein concentration")
plt.title("Protein Production and Removal")
plt.tight_layout()
plt.show()
````

## 20. How Should We Interpret the Graph?

At the beginning, protein concentration is low.

Protein is being produced faster than it is removed, so the concentration increases.

Later, removal becomes more important as the protein concentration increases.

Eventually, production and removal become approximately balanced.

The concentration then approaches a **steady state**.


## 21. What is a Steady State?

A steady state occurs when the quantity is no longer changing significantly.

In ODE language:

$$\frac{dP}{dt}=0$$

For our model:

$$k_{prod}-k_{deg}P=0$$

The important biological idea is simply:

> **Production = Removal**

At this point, the protein concentration remains approximately constant.


## 22. Connecting Back to the Signaling Model

We can now understand the two-variable signaling model:

$$\frac{dA}{dt}=k_1S-k_2A$$

$$\frac{dR}{dt}=k_3A-k_4R$$

The computer solves both equations together.

The result tells us how **activated protein** and **cellular response** change over time.
